Clean and standardize order_date

```python

import pandas as pd
import re

def clean_order_date(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    formats = [
        "%d/%m/%Y",
        "%d-%m-%y",
        "%Y-%m-%d",
        "%d-%m-%Y",
        "%m/%d/%Y"
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(value, format=fmt, errors="raise")
        except ValueError:
            continue

    return pd.NaT


for year, df in transactions.items():
    df["order_date_clean"] = df["order_date"].apply(clean_order_date)
    df["order_date_clean"] = df["order_date_clean"].dt.strftime("%Y-%m-%d")

    print(
        year,
        "| Invalid or missing dates:",
        df["order_date_clean"].isna().sum()
    )

print("\nSample date comparison:")
display(
    transactions[2025][["order_date", "order_date_clean"]].head(10)
)
```

Question 2 — Clean original_price_inr

```python

import numpy as np
import pandas as pd
import re

def clean_price(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value.lower() in ["", "price on request", "nan", "none", "n/a"]:
        return np.nan

    value = value.replace("₹", "").replace(",", "").replace(" ", "")
    value = re.sub(r"[^0-9.\-]", "", value)

    try:
        return float(value) if value else np.nan
    except ValueError:
        return np.nan


for year, df in transactions.items():
    df["original_price_inr_clean"] = df["original_price_inr"].apply(clean_price)

    print(
        year,
        "| Missing or unparseable prices:",
        df["original_price_inr_clean"].isna().sum()
    )

print("\nSample price comparison:")
display(
    transactions[2025][
        ["original_price_inr", "original_price_inr_clean"]
    ].head(10)
)
```

Question 3 — Standardize customer ratings

```python

import pandas as pd
import numpy as np
import re

def clean_rating(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    if value in ["", "nan", "none", "n/a"]:
        return np.nan

    match = re.fullmatch(r"(\d+(?:\.\d+)?)\s*/\s*(\d+(?:\.\d+)?)", value)

    if match:
        rating = float(match.group(1))
        scale = float(match.group(2))

        if scale <= 0:
            return np.nan

        rating = rating * 5 / scale
    else:
        match = re.search(r"\d+(?:\.\d+)?", value)

        if not match:
            return np.nan

        rating = float(match.group())

    if 1 <= rating <= 5:
        return round(rating, 2)

    return np.nan


for year, df in transactions.items():
    df["customer_rating_clean"] = df["customer_rating"].apply(clean_rating)

    print(
        year,
        "| Missing or invalid ratings:",
        df["customer_rating_clean"].isna().sum()
    )

print("\nSample rating comparison:")
display(
    transactions[2025][
        ["customer_rating", "customer_rating_clean"]
    ].head(10)
)
```

Question 7 — Clean delivery days

```python
def clean_delivery_days(value):
    if pd.isna(value):
        return np.nan

    text = str(value).strip().lower()

    if text in ["same day", "same-day", "same day delivery"]:
        return 0

    numbers = re.findall(r"\d+(?:\.\d+)?", text)

    if not numbers:
        return np.nan

    values = [float(n) for n in numbers]

    if "-" in text and len(values) >= 2:
        days = sum(values[:2]) / 2
    else:
        days = values[0]

    if days < 0 or days > 30:
        return np.nan

    return days

for year, df in transactions.items():
    df["delivery_days_clean"] = df["delivery_days"].apply(clean_delivery_days)

print("Delivery day comparison:")
display(
    transactions[2025][["delivery_days", "delivery_days_clean"]]
    .drop_duplicates()
    .head(30)
)

print("\nInvalid or missing delivery values by year:")
for year, df in transactions.items():
    print(year, df["delivery_days_clean"].isna().sum())
```

Question 3 — Customer segmentation using RFM

```python

import pandas as pd
import matplotlib.pyplot as plt

all_transactions = pd.concat(
    transactions.values(),
    ignore_index=True
)

all_transactions["order_date_clean"] = pd.to_datetime(
    all_transactions["order_date_clean"],
    errors="coerce"
)

all_transactions["final_amount_inr"] = pd.to_numeric(
    all_transactions["final_amount_inr"],
    errors="coerce"
)

reference_date = all_transactions["order_date_clean"].max() + pd.Timedelta(days=1)

rfm = all_transactions.dropna(
    subset=["customer_id", "order_date_clean"]
).groupby("customer_id").agg(
    Recency=("order_date_clean", lambda x: (reference_date - x.max()).days),
    Frequency=("transaction_id", "nunique"),
    Monetary=("final_amount_inr", "sum")
)

rfm["R_score"] = pd.qcut(
    rfm["Recency"].rank(method="first", ascending=True),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

rfm["F_score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm["M_score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

def assign_segment(row):
    if row["R_score"] >= 3 and row["F_score"] >= 3 and row["M_score"] >= 3:
        return "Champions"
    elif row["R_score"] >= 3 and row["F_score"] >= 2:
        return "Loyal Customers"
    elif row["R_score"] <= 2 and row["F_score"] >= 3:
        return "At Risk"
    elif row["R_score"] >= 3 and row["F_score"] == 1:
        return "New Customers"
    else:
        return "Needs Attention"

rfm["Segment"] = rfm.apply(assign_segment, axis=1)

plt.figure(figsize=(10, 6))
scatter = plt.scatter(
    rfm["Frequency"],
    rfm["Monetary"],
    c=rfm["Recency"],
    alpha=0.6
)
plt.colorbar(scatter, label="Recency (days)")
plt.title("Customer RFM Segmentation")
plt.xlabel("Purchase Frequency")
plt.ylabel("Total Monetary Value (INR)")
plt.tight_layout()
plt.show()

print("Customers by segment:")
print(rfm["Segment"].value_counts())

display(
    rfm.groupby("Segment")[["Recency", "Frequency", "Monetary"]]
    .mean()
    .round(2)
)
```

```python

festival_data = all_transactions.copy()

festival_data["order_date_clean"] = pd.to_datetime(
    festival_data["order_date_clean"], errors="coerce"
)
festival_data["final_amount_inr"] = pd.to_numeric(
    festival_data["final_amount_inr"], errors="coerce"
)

festival_data = festival_data.dropna(
    subset=["order_date_clean", "final_amount_inr"]
)

festival_data["festival_name"] = (
    festival_data["festival_name"]
    .astype("string")
    .str.strip()
)

festival_data["festival_name"] = festival_data["festival_name"].replace(
    ["", "nan", "None", "<NA>"], pd.NA
)

festival_data["festival_flag"] = (
    festival_data["is_festival_sale_clean"]
    .astype("string")
    .str.lower()
    .str.strip()
)

festival_data["festival_flag"] = festival_data["festival_flag"].map({
    "true": True, "false": False,
    "yes": True, "no": False,
    "1": True, "0": False
})

festival_data["month"] = festival_data["order_date_clean"].dt.to_period("M").dt.to_timestamp()

monthly_revenue = (
    festival_data.groupby("month")["final_amount_inr"]
    .sum()
    .sort_index()
)

plt.figure(figsize=(14, 6))
plt.plot(monthly_revenue.index, monthly_revenue.values, marker="o")
plt.title("Monthly Revenue Trend and Festival Sales Impact")
plt.xlabel("Month")
plt.ylabel("Revenue (INR)")
plt.ticklabel_format(axis="y", style="plain")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

festival_monthly = (
    festival_data.groupby(["month", "festival_flag"])["final_amount_inr"]
    .sum()
    .unstack(fill_value=0)
)

festival_monthly.plot(
    kind="area", stacked=True, figsize=(14, 6), alpha=0.75
)
plt.title("Monthly Revenue: Festival vs Non-Festival Sales")
plt.xlabel("Month")
plt.ylabel("Revenue (INR)")
plt.ticklabel_format(axis="y", style="plain")
plt.tight_layout()
plt.show()

festival_revenue = (
    festival_data.dropna(subset=["festival_name"])
    .groupby("festival_name")
    .agg(
        revenue=("final_amount_inr", "sum"),
        orders=("transaction_id", "nunique"),
        customers=("customer_id", "nunique")
    )
    .sort_values("revenue", ascending=False)
)

display(festival_revenue.round(2))

if not festival_revenue.empty:
    festival_revenue["revenue"].sort_values().plot(
        kind="barh", figsize=(10, 6)
    )
    plt.title("Revenue by Festival")
    plt.xlabel("Revenue (INR)")
    plt.ylabel("Festival")
    plt.ticklabel_format(axis="x", style="plain")
    plt.tight_layout()
    plt.show()

festival_data["festival_window"] = "No Festival"

festival_data.loc[
    festival_data["festival_flag"].eq(True), "festival_window"
] = "During Festival"

festival_dates = (
    festival_data.loc[
        festival_data["festival_flag"].eq(True)
        & festival_data["festival_name"].notna(),
        ["festival_name", "order_date_clean"]
    ]
    .drop_duplicates()
)

window_results = []

for festival, dates in festival_dates.groupby("festival_name"):
    for festival_date in dates["order_date_clean"]:
        start = festival_date.normalize()
        before_start = start - pd.Timedelta(days=7)
        after_end = start + pd.Timedelta(days=7)

        before = festival_data[
            (festival_data["order_date_clean"] >= before_start)
            & (festival_data["order_date_clean"] < start)
        ]

        during = festival_data[
            (festival_data["order_date_clean"] >= start)
            & (festival_data["order_date_clean"] < start + pd.Timedelta(days=1))
            & (festival_data["festival_name"] == festival)
        ]

        after = festival_data[
            (festival_data["order_date_clean"] >= start + pd.Timedelta(days=1))
            & (festival_data["order_date_clean"] < after_end)
        ]

        window_results.append({
            "festival": festival,
            "festival_date": start,
            "before_daily_avg_revenue": before["final_amount_inr"].sum() / 7,
            "during_revenue": during["final_amount_inr"].sum(),
            "after_daily_avg_revenue": after["final_amount_inr"].sum() / 7
        })

festival_impact = pd.DataFrame(window_results)

if not festival_impact.empty:
    festival_impact = festival_impact.drop_duplicates(
        subset=["festival", "festival_date"]
    )
    display(festival_impact.round(2))

    impact_plot = festival_impact.groupby("festival")[
        [
            "before_daily_avg_revenue",
            "during_revenue",
            "after_daily_avg_revenue"
        ]
    ].mean()

    impact_plot.plot(kind="bar", figsize=(12, 6))
    plt.title("Average Revenue Before, During, and After Festivals")
    plt.xlabel("Festival")
    plt.ylabel("Average Revenue (INR)")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

Question 14 — Customer lifetime value and cohort retention

```python

clv_data = all_transactions.copy()

clv_data["order_date_clean"] = pd.to_datetime(
    clv_data["order_date_clean"], errors="coerce"
)
clv_data["final_amount_inr"] = pd.to_numeric(
    clv_data["final_amount_inr"], errors="coerce"
)

clv_data = clv_data.dropna(
    subset=["customer_id", "order_date_clean", "final_amount_inr"]
)

clv_data["order_month"] = (
    clv_data["order_date_clean"].dt.to_period("M")
)

customer_clv = (
    clv_data.groupby("customer_id")
    .agg(
        lifetime_value=("final_amount_inr", "sum"),
        order_count=("transaction_id", "nunique"),
        first_order=("order_date_clean", "min"),
        last_order=("order_date_clean", "max")
    )
)

customer_clv["acquisition_year"] = customer_clv["first_order"].dt.year
customer_clv["customer_lifetime_days"] = (
    customer_clv["last_order"] - customer_clv["first_order"]
).dt.days

display(customer_clv.describe().round(2))

customer_clv["lifetime_value"].plot(
    kind="hist", bins=50, figsize=(10, 5)
)
plt.title("Customer Lifetime Value Distribution")
plt.xlabel("Lifetime Value (INR)")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()

clv_segment = (
    clv_data.groupby("customer_spending_tier")
    .agg(
        average_customer_value=("final_amount_inr", "mean"),
        total_revenue=("final_amount_inr", "sum"),
        orders=("transaction_id", "nunique"),
        customers=("customer_id", "nunique")
    )
)

clv_segment["revenue_per_customer"] = (
    clv_segment["total_revenue"] / clv_segment["customers"]
)

display(clv_segment.round(2))

clv_segment["revenue_per_customer"].sort_values().plot(
    kind="barh", figsize=(9, 5)
)
plt.title("Revenue per Customer by Spending Tier")
plt.xlabel("Revenue per Customer (INR)")
plt.tight_layout()
plt.show()

customer_month = (
    clv_data.groupby(["customer_id", "order_month"])["final_amount_inr"]
    .sum().reset_index()
)

cohort_info = (
    clv_data.groupby("customer_id")["order_month"]
    .min().rename("cohort_month")
)

customer_month = customer_month.join(
    cohort_info, on="customer_id"
)

customer_month["cohort_index"] = (
    (customer_month["order_month"].dt.year -
     customer_month["cohort_month"].dt.year) * 12
    + customer_month["order_month"].dt.month
    - customer_month["cohort_month"].dt.month
)

cohort_counts = customer_month.groupby(
    ["cohort_month", "cohort_index"]
)["customer_id"].nunique().unstack()

cohort_sizes = cohort_counts.iloc[:, 0]
retention = cohort_counts.div(cohort_sizes, axis=0) * 100

display(retention.round(1))

plt.figure(figsize=(14, 8))
sns.heatmap(retention, cmap="YlGnBu", annot=False, fmt=".0f")
plt.title("Monthly Customer Cohort Retention (%)")
plt.xlabel("Months Since First Purchase")
plt.ylabel("Acquisition Cohort")
plt.tight_layout()
plt.show()

retention.mean(axis=0).plot(
    marker="o", figsize=(10, 5)
)
plt.title("Average Customer Retention Curve")
plt.xlabel("Months Since First Purchase")
plt.ylabel("Average Retention (%)")
plt.tight_layout()
plt.show()

customer_clv.groupby("acquisition_year")["lifetime_value"].mean().plot(
    marker="o", figsize=(10, 5)
)
plt.title("Average Customer Lifetime Value by Acquisition Year")
plt.xlabel("Acquisition Year")
plt.ylabel("Average Lifetime Value (INR)")
plt.tight_layout()
plt.show()
```

Question 17 — Customer journey and purchase transitions

```python

journey_data = all_transactions.copy()

journey_data["order_date_clean"] = pd.to_datetime(
    journey_data["order_date_clean"], errors="coerce"
)

journey_data = journey_data.dropna(
    subset=["customer_id", "order_date_clean"]
).sort_values(["customer_id", "order_date_clean"])

journey_data["purchase_number"] = (
    journey_data.groupby("customer_id").cumcount() + 1
)

journey_data["first_purchase_date"] = (
    journey_data.groupby("customer_id")["order_date_clean"].transform("min")
)

journey_data["days_since_first_purchase"] = (
    journey_data["order_date_clean"] - journey_data["first_purchase_date"]
).dt.days

customer_journey = (
    journey_data.groupby("customer_id")
    .agg(
        first_purchase=("order_date_clean", "min"),
        last_purchase=("order_date_clean", "max"),
        purchase_count=("transaction_id", "nunique"),
        total_spending=("final_amount_inr", "sum")
    )
)

customer_journey["customer_stage"] = pd.cut(
    customer_journey["purchase_count"],
    bins=[0, 1, 3, 10, float("inf")],
    labels=["One-time", "Occasional", "Repeat", "Loyal"]
)

display(customer_journey.groupby(
    "customer_stage", observed=False
).agg(
    customers=("purchase_count", "count"),
    average_purchases=("purchase_count", "mean"),
    average_spending=("total_spending", "mean")
).round(2))

customer_journey["customer_stage"].value_counts().plot(
    kind="bar", figsize=(9, 5)
)
plt.title("Customer Distribution by Purchase Stage")
plt.xlabel("Customer Stage")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()

journey_data["previous_category"] = (
    journey_data.groupby("customer_id")["category_clean"].shift(1)
)

transition_data = journey_data.dropna(
    subset=["previous_category", "category_clean"]
)

transition_matrix = pd.crosstab(
    transition_data["previous_category"],
    transition_data["category_clean"],
    normalize="index"
) * 100

display(transition_matrix.round(2))

plt.figure(figsize=(12, 9))
sns.heatmap(transition_matrix, cmap="Blues", annot=True, fmt=".1f")
plt.title("Category Transition Matrix (%)")
plt.xlabel("Next Category")
plt.ylabel("Previous Category")
plt.tight_layout()
plt.show()

top_transitions = (
    transition_data.groupby(["previous_category", "category_clean"])
    .size().reset_index(name="transition_count")
    .sort_values("transition_count", ascending=False)
    .head(15)
)

display(top_transitions)

top_transitions["transition"] = (
    top_transitions["previous_category"].astype(str)
    + " → "
    + top_transitions["category_clean"].astype(str)
)

top_transitions.sort_values("transition_count").plot(
    x="transition", y="transition_count",
    kind="barh", legend=False, figsize=(11, 7)
)
plt.title("Most Common Customer Category Transitions")
plt.xlabel("Transition Count")
plt.ylabel("Category Transition")
plt.tight_layout()
plt.show()

customer_journey["acquisition_year"] = customer_journey["first_purchase"].dt.year

display(customer_journey.groupby("acquisition_year").agg(
    customers=("purchase_count", "count"),
    average_purchases=("purchase_count", "mean"),
    average_lifetime_spending=("total_spending", "mean")
).round(2))
```

```python
#Join transactions, products, customers, and time dimension
join_query = """
SELECT
    td.year,
    p.category,
    p.subcategory,
    COUNT(t.transaction_id) AS total_transactions,
    SUM(t.quantity) AS units_sold,
    ROUND(SUM(t.final_amount_inr), 2) AS total_revenue_inr,
    ROUND(AVG(t.final_amount_inr), 2) AS average_transaction_value,
    COUNT(DISTINCT c.customer_id) AS unique_customers
FROM transactions t
JOIN products p
    ON t.product_id = p.product_id
JOIN customers c
    ON t.customer_id = c.customer_id
JOIN time_dimension td
    ON t.order_date = td.date_key
GROUP BY
    td.year,
    p.category,
    p.subcategory
ORDER BY
    td.year,
    total_revenue_inr DESC
"""

category_year_df = pd.read_sql_query(join_query, conn)

display(category_year_df.head(15))
```

```python
APP_DIR = Path(r"C:\nandish\sam projects\AmazonIndiaProject")
APP_PATH = APP_DIR / "app.py"

dashboard_code = '''
import sqlite3
from pathlib import Path

import pandas as pd
import streamlit as st

DB_PATH = Path(r"C:\\nandish\\sam projects\\AmazonIndiaProject\\data\\amazon_india_analytics.db")

st.set_page_config(page_title="Amazon India Sales Analytics", layout="wide")
st.title("Amazon India: A Decade of Sales Analytics")
st.caption("Sales performance dashboard | 2015–2025")

@st.cache_data
def load_data():
    with sqlite3.connect(DB_PATH) as conn:
        kpis = pd.read_sql_query("""
            SELECT
                SUM(final_amount_inr) AS revenue,
                COUNT(*) AS transactions,
                SUM(quantity) AS units,
                COUNT(DISTINCT customer_id) AS customers
            FROM transactions
        """, conn)

        yearly = pd.read_sql_query("""
            SELECT order_year AS year,
                   SUM(final_amount_inr) AS revenue,
                   COUNT(*) AS transactions
            FROM transactions
            GROUP BY order_year
            ORDER BY order_year
        """, conn)

        categories = pd.read_sql_query("""
            SELECT category,
                   SUM(final_amount_inr) AS revenue
            FROM transactions
            GROUP BY category
            ORDER BY revenue DESC
        """, conn)

        payments = pd.read_sql_query("""
            SELECT payment_method,
                   SUM(final_amount_inr) AS revenue
            FROM transactions
            GROUP BY payment_method
            ORDER BY revenue DESC
        """, conn)

    return kpis, yearly, categories, payments

kpis, yearly, categories, payments = load_data()

col1, col2, col3, col4 = st.columns(4)
col1.metric("Total Revenue", f"₹{kpis.loc[0, 'revenue']:,.0f}")
col2.metric("Transactions", f"{kpis.loc[0, 'transactions']:,.0f}")
col3.metric("Units Sold", f"{kpis.loc[0, 'units']:,.0f}")
col4.metric("Unique Customers", f"{kpis.loc[0, 'customers']:,.0f}")

st.subheader("Yearly Revenue Trend")
st.line_chart(yearly.set_index("year")["revenue"])

st.subheader("Yearly Transaction Volume")
st.bar_chart(yearly.set_index("year")["transactions"])

st.subheader("Revenue by Category")
st.bar_chart(categories.set_index("category")["revenue"])

st.subheader("Revenue by Payment Method")
st.bar_chart(payments.set_index("payment_method")["revenue"])

st.subheader("Yearly Revenue Data")
st.dataframe(yearly, use_container_width=True)

st.subheader("Category Performance")
st.dataframe(categories, use_container_width=True)
'''

APP_PATH.write_text(dashboard_code, encoding="utf-8")

print("Dashboard file created:", APP_PATH)
print("File exists:", APP_PATH.exists())
```

```python
#Launch the dashboard
import subprocess
import sys

process = subprocess.Popen(
    [sys.executable, "-m", "streamlit", "run", str(APP_PATH)],
    cwd=str(APP_DIR)
)

print("Streamlit process started.")
print("Open the local URL shown in the output, usually http://localhost:8501")
```
